#### Imports

```python
import pandas as pd
```

#### Load Database

```python
event =  'elections'
```

```python
df = pd.read_parquet(f'../../NLP Pipeline/elections_reingest_annotated.parquet')

df = df.drop(columns=['Unnamed: 0'])
# missed removing index column while making the CSVs, sorry!

df
```

#### Set Up Environment

```python
synonyms = ['opinion', 'editorial', 'column', 'op-ed', 'op ed', 'analysis']
```

```python
def check_presence(text):
    text = str(text).lower()
    return any(synonym in text for synonym in synonyms)
```

```python
df_opinions = df[df['title'].apply(check_presence) | df['url'].apply(check_presence)]

df_opinions
```

```python
df_mainstream = df.drop(df_opinions.index)

df_mainstream
```

```python
df_opinions.to_parquet(f'../../NLP Pipeline/{event}_opinions_reingest_annotated.parquet', index=False)
df_mainstream.to_parquet(f'../../NLP Pipeline/{event}_mainstream_reingest_annotated.parquet', index=False)
```
